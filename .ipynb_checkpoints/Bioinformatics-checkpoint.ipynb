<b>What is Bioinformatics? 
<br>Google says...</b><br>
<p>
"Bioinformatics is an interdisciplinary field that merges biology, computer science, statistics, and information technology to analyze, interpret, and manage complex biological data, such as DNA, RNA, and protein sequences. It is critical for genomics, drug discovery, and molecular modeling, enabling scientists to uncover patterns and gain insights into molecular function."</p>

<b>Put another way...</b><br>
<p>Bioinformatics is essentially BIG data analysis for biological data sets. 
It requires computational and statistical analyses in order to extract meaning from biological data. 
Since this is the case, bioinformatics can also be used to refer to the development of the software and methods which are used to understand biological data...</p>

## The Central Dogma of Molecular Biology:
DNA->RNA-> PROTEIN
<p>
DNA is transcribed into RNA, which is then translated into protein. This fundamental process ensures that instructions in the genome are executed to build cellular structures and perform functions.

# RNA Alphabet ( the T are replaced by U ) on messenger RNA
## A,U,G,C
<pre>  
DNA
  --T--A--T--G--T--C
    +  +  +  +  +  +

## Protein Alphabet has 20 letters, each is an amino acid

### 'A', 'C', 'D', 'E', 'F', 'G', 'H', 'I', 'K', 'L', 'M', 'N', 'P', 'Q', 'R', 'S', 'T', 'V', 'W', 'Y' 

##  The genetic code consists of 64 triplets (codons) of nucleotides.
### 61 coding for 20 amino acids and 3 acting as stop signals.
### Multiple codons can specify the same amino acid. 


- What amino acid gets created from codon <b>AUA </b>?
- What amino acid gets created from codon <b>CAG </b>?

Standard Amino Acid Codon List

    Alanine (Ala, A): GCU, GCC, GCA, GCG
    Arginine (Arg, R): CGU, CGC, CGA, CGG, AGA, AGG
    Asparagine (Asn, N): AAU, AAC
    Aspartic acid (Asp, D): GAU, GAC
    Cysteine (Cys, C): UGU, UGC
    Glutamine (Gln, Q): CAA, CAG*
    Glutamic acid (Glu, E): GAA, GAG
    Glycine (Gly, G): GGU, GGC, GGA, GGG
    Histidine (His, H): CAU, CAC
    Isoleucine (Ile, I): AUU, AUC, AUA*  
    Leucine (Leu, L): UUA, UUG, CUU, CUC, CUA, CUG
    Lysine (Lys, K): AAA, AAG
    Methionine (Met, M) [Start]: AUG
    Phenylalanine (Phe, F): UUU, UUC
    Proline (Pro, P): CCU, CCC, CCA, CCG
    Serine (Ser, S): UCU, UCC, UCA, UCG, AGU, AGC
    Threonine (Thr, T): ACU, ACC, ACA, ACG
    Tryptophan (Trp, W): UGG
    Tyrosine (Tyr, Y): UAU, UAC
    Valine (Val, V): GUU, GUC, GUA, GUG 
        
Stop sequences: <b>UAA,UAG,UGA</b>

```python
<code>
# Example python code to generate condons or kmer
def generate_trimers():
    bases = ['A','U','G','C']
    result = []
    for base1 in bases:
        for base2 in bases:
            for base3 in bases:
                result.append( base1 + base2 + base3 )
    return result
</code>
```

<code>
# Using recursion, and create kmer of any length 
def generate_kmers_rec(length):
    if length == 1:
        return ['A','U','G','C']
    else:
        result = []
        for seq in generate_kmers_rec( length - 1 ):
            for base in ['A','U','G','C']:
                result.append( seq + base )
        return result
</code>

```python
# Using recursion, and create kmer of any length. 
# Cutting long strings into kmer is useful for training AI models.
def generate_kmers_rec(length):
    if length == 1:
        return ['A','U','G','C']
    else:
        result = []
        for seq in generate_kmers_rec( length - 1 ):
            for base in ['A','U','G','C']:
                result.append( seq + base )
        return result

rna = generate_kmers_rec(3)
print( *rna )
```

# AI and Protein folding : Alphafold
AlphaFold is
a revolutionary artificial intelligence system developed by Google DeepMind that predicts a protein's 3D structure from its 1D amino acid sequence with near-experimental accuracy. By solving the 50-year-old "protein folding problem," AlphaFold has transformed structural biology, accelerating research in drug discovery, disease understanding, and biotechnology.
